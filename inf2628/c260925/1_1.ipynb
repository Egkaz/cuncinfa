которая принимает произвольное количество кортежей вида
(название_пакета, количество_услуг, базовая_стоимость)
и возвращает СУММУ базовых стоимостей тех пакетов,
у которых количество_услуг >= 4.

Используйте генератор списка с условием и sum().

Данные для вызова:
  ('Demo', 1, 2000)
  ('Basic', 4, 22000)
  ('Plus', 6, 38000)
  ('Pro', 9, 72000)
  ('Ultra', 13, 105000)
  ('Micro', 2, 7000)

Вызов: filter_packages(('Demo', 1, 2000), ('Basic', 4, 22000), ('Plus', 6, 38000), ('Pro', 9, 72000), ('Ultra', 13, 105000), ('Micro', 2, 7000))

```python
def filter_packages(*packages, min_services=4):
    res = 0
    b = []
    for el in packages:
        if el[1] >= min_services:
            b.append(el[2])
    return sum(b)


def filter_packages(*packages, min_services=4):
    return sum([el[2] for el in packages if el[1] >= min_services])


filter_packages(('Demo', 1, 2000), ('Basic', 4, 22000), 
                ('Plus', 6, 38000), ('Pro', 9, 72000), 
                ('Ultra', 13, 105000), ('Micro', 2, 7000))
```

```python
class FibonacciIterator:
    def __init__(self, stop=10):
        self._stop = stop
        self._index = 0
        
        self._current = 0
        self._next = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self._index < self._stop:
            self._index += 1
            fib_number = self._current
            self._current, self._next = (
                self._next,
                self._current + self._next,
            )
            return fib_number
        else:
            raise StopIteration
```

```python
fi = FibonacciIterator()
isinstance(fi, Iterable), isinstance(fi, Iterator)
```

```python
class FibonacciIterator:
    def __init__(self, stop=10):
        self._stop = stop
        self._index = 0
        self._current = 0
        self._next = 1

    def __next__(self):
        if self._index < self._stop:
            self._index += 1
            fib_number = self._current
            self._current, self._next = (
                self._next,
                self._current + self._next,
            )
            return fib_number
        else:
            raise StopIteration

class FibonacciIterable:
    def __iter__(self):
        return FibonacciIterator()
```

```python
fiterable = FibonacciIterable()
```
