## Урок 2. Условия и циклы

### Теория: условия и циклы

> **Условия**
> - `if условие:` выполняет блок кода, если условие истинно.
> - `elif` проверяет следующее условие.
> - `else` выполняется, если ни одно условие не подошло.
> - Для сравнения используются `==`, `!=`, `>`, `<`, `>=`, `<=`.
> - Логические операторы: `and`, `or`, `not`.

> **Циклы**
> - `while` повторяет код, пока условие истинно.
> - `for` удобно использовать вместе с `range()`.
> - `break` полностью останавливает цикл.
> - `continue` пропускает текущую итерацию и переходит к следующей.

Примеры диапазонов:

- `range(5)` — числа 0, 1, 2, 3, 4;
- `range(1, 5)` — числа 1, 2, 3, 4;
- `range(1, 10, 2)` — числа 1, 3, 5, 7, 9.

````python
# Пример 1. Простое условие
age = int(input("Сколько тебе лет? "))

if age >= 18:
    print("Доступ разрешён")
else:
    print("Доступ запрещён")
````

````python
# Пример 2. Несколько условий через elif
score = int(input("Введи балл от 0 до 100: "))

if score >= 90:
    print("Оценка: 5")
elif score >= 75:
    print("Оценка: 4")
elif score >= 60:
    print("Оценка: 3")
else:
    print("Оценка: 2")
````

````python
# Пример 3. Логические операторы
login = input("Логин: ")
password = input("Пароль: ")

if login == "admin" and password == "12345":
    print("Вход выполнен")
else:
    print("Неверный логин или пароль")
````

````python
# Пример 4. Цикл while
counter = 1

while counter <= 5:
    print("Счётчик:", counter)
    counter += 1

print("Цикл завершён")
````

````python
numbers = [1, 3, 5, 7, 9]
target = 4
i = 0

while i < len(numbers):
    if numbers[i] == target:
        print(f"Число {target} найдено на позиции {i}")
        break  # Досрочное прерывание
    i += 1
else:
    # Этот блок выполнится, только если цикл завершился, потому что условие `i < len(numbers)` стало ложным
    print(f"Число {target} не найдено в списке")
````

````python
# Пример 5. Цикл for и сумма чисел
total = 0

for i in range(1, 6):
    print("Шаг:", i)
    total += i

print("Сумма от 1 до 5:", total)
````

````python
# Пример 6. continue и break
for i in range(1, 11):
    if i == 3:
        continue

    if i == 8:
        break

    print(i)
````

## Урок 3. Списки, словари, кортежи, множества и функции

### Теория: коллекции и функции

> **Списки**
> - Список создаётся так: `numbers = [1, 2, 3]`.
> - Индексация начинается с нуля: `numbers[0]`.
> - Добавить элемент: `numbers.append(4)`.
> - Длина списка: `len(numbers)`.

> **Словари**
> - Словарь хранит пары ключ-значение.
> - Пример: `user = {"name": "Аня", "age": 16}`.
> - Получить значение: `user["name"]`.
> - Добавить значение: `user["city"] = "Москва"`.

> **Кортежи и множества**
> - Кортеж похож на список, но его нельзя изменять: `point = (10, 20)`.
> - Множество хранит только уникальные элементы: `set([1, 1, 2])`.

> **Функции**
> - Функция создаётся через `def`.
> - Она может принимать параметры.
> - Она может возвращать значение через `return`.
