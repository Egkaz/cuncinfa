def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Ошибка: деление на ноль"
    return a / b

x = float(input("Первое число: "))
operation = input("Операция (+, -, *, /): ")
y = float(input("Второе число: "))

if operation == "+":
    result = add(x, y)
elif operation == "-":
    result = subtract(x, y)
elif operation == "*":
    result = multiply(x, y)
elif operation == "/":
    result = divide(x, y)
else:
    result = "Неизвестная операция"

print("Результат:", result)
