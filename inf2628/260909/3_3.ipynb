## Урок 4. Строки, срезы, методы и углублённая работа со списками

### Теория: строки, срезы и списки

> **Индексы**
> - Элементы строк и списков нумеруются с нуля.
> - `text[0]` — первый элемент.
> - `text[-1]` — последний элемент.

> **Срезы**
> - Формат: `объект[начало:конец:шаг]`.
> - Элемент с индексом `конец` не включается.
> - `text[::-1]` переворачивает строку или список.

> **Методы строк**
> - `.strip()` убирает пробелы по краям.
> - `.lower()` переводит в нижний регистр.
> - `.upper()` переводит в верхний регистр.
> - `.replace()` заменяет подстроку.
> - `.find()` ищет подстроку.
> - `.split()` разбивает строку на список.
> - `.join()` собирает список в строку.

> **Списки**
> - `.sort()` сортирует список на месте.
> - `sorted()` возвращает новый отсортированный список.
> - Вложенные списки удобно обходить циклом `for`.
> - `enumerate()` даёт номер элемента и сам элемент.

````python
# Пример 1. Индексы и срезы
text = "Программирование"

print(text[0])
print(text[1])
print(text[-1])

print(text[0:4])
print(text[:4])
print(text[4:])

print(text[::-1])
````

````python
# Пример 2. Методы строк
message = "  Привет, Python!  "

print(message.strip())
print(message.lower())
print(message.upper())
print(message.replace("Python", "мир"))
print(len(message))
````

````python
# Пример 3. Поиск в строке
text = "мама мыла раму"

print(text.find("мыла"))
print(text.count("а"))
print(text.startswith("мама"))
print(text.endswith("раму"))
````

````python
# Пример 4. split и join
line = "яблоко,банан,апельсин"

fruits = line.split(",")

print(fruits)
print(fruits[0])
print(fruits[2])

new_line = " | ".join(fruits)

print(new_line)
````

````python
# Пример 5. Сортировка списков
numbers = [3, 1, 4, 1, 5, 9, 2, 6]

print(sorted(numbers))
print(numbers)

numbers.sort()
print(numbers)

numbers.sort(reverse=True)
print(numbers)
````

````python
# Пример 6. Вложенные списки и enumerate
students = [
    ["Аня", 5],
    ["Боря", 4],
    ["Вера", 5]
]

for index, student in enumerate(students, start=1):
    name, grade = student
    print(f"{index}. {name} - {grade}")
````

## Урок 5. Функции глубже, исключения, модули и файлы

### Теория: функции, исключения, модули и файлы

> **Функции**
> - У параметров могут быть значения по умолчанию.
> - Аргументы можно передавать по имени.
> - Функция может возвращать несколько значений через запятую.

> **Исключения**
> - `try` выполняет код, который может вызвать ошибку.
> - `except` обрабатывает ошибку.
> - `else` выполняется, если ошибки не было.
> - `finally` выполняется всегда.

> **Модули**
> - `import math` подключает математический модуль.
> - `import random` подключает модуль случайных чисел.

> **Файлы**
> - Режим `"r"` — чтение.
> - Режим `"w"` — запись, стирает старое содержимое.
> - Режим `"a"` — дозапись в конец файла.
> - Удобно использовать конструкцию `with open(...) as file:`.

````python
# Пример 1. Параметры по умолчанию и именованные аргументы
def create_profile(name, age, city="Неизвестно"):
    print(f"Имя: {name}, Возраст: {age}, Город: {city}")

create_profile("Аня", 16)
create_profile("Боря", 17, "Москва")
create_profile(city="Казань", name="Вера", age=15)
````

````python
# Пример 2. Функция возвращает несколько значений
def get_min_and_max(numbers):
    return min(numbers), max(numbers)

data = [3, 7, 1, 9, 4]

minimum, maximum = get_min_and_max(data)

print("Минимум:", minimum)
print("Максимум:", maximum)
````

````python
# Пример 3. Обработка ошибок
try:
    x = int(input("Делимое: "))
    y = int(input("Делитель: "))

    result = x / y

except ValueError:
    print("Нужно ввести целое число")

except ZeroDivisionError:
    print("Нельзя делить на ноль")

else:
    print("Результат:", result)

finally:
    print("Операция завершена")
````

````python
# Пример 4. Функция проверки пароля
def is_strong_password(password, min_length=8):
    if len(password) < min_length:
        return False

    has_digit = False
    has_upper = False
    has_lower = False

    for ch in password:
        if ch.isdigit():
            has_digit = True
        elif ch.isupper():
            has_upper = True
        elif ch.islower():
            has_lower = True

    return has_digit and has_upper and has_lower

print(is_strong_password("Qwerty123"))
print(is_strong_password("qwerty"))
````

````python
# Пример 5. Модули math и random
import math
import random

print(math.pi)
print(math.sqrt(16))

print(random.randint(1, 10))
````

````python
# Пример 6. Запись, дозапись и чтение файла
file_name = "notes.txt"

with open(file_name, "w", encoding="utf-8") as file:
    file.write("Первая запись")

with open(file_name, "a", encoding="utf-8") as file:
    file.write("\nВторая запись")

with open(file_name, "r", encoding="utf-8") as file:
    print(file.read())
````
