````python
from typing import Self

class Vector:
    def __init__(self, x: float, y: float) -> None:
        self.x: float = x
        self.y: float = y

    def __add__(self, obj: Self) -> Self:
        # return Vector(self.x + obj.x, self.y + obj.y)
        # return type(self)(self.x + obj.x, self.y + obj.y)
        return self.__class__(self.x + obj.x, self.y + obj.y)

    def __repr__(self) -> str:
        return f"({self.x}, {self.y})"
````

#### Вектор со любым количеством координат

````python
class VectorDimensionsDoNotMatch(Exception):
    """
    Ошибка: Размерности векторов не совпадают
    """
    def __init__(self, l1, l2):
        message = f"Размерности векторов не одинаковы {l1} != {l2}"
        super().__init__(message)
        

class Vector:
    def __init__(self, *args):
        """
        x - координаты вектора
        """
        self.x = args

    def __add__(self: Self, obj: Self) -> Self:
        if len(self.x) != len(obj.x):
            raise VectorDimensionsDoNotMatch(len(self.x), len(obj.x))
        else:
            a = []
            for i in range(len(self.x)):
                a.append(self.x[i] + obj.x[i])
            return type(self)(*a)
            
    def __repr__(self: Self) -> str:
        return f"({", ".join(map(str, self.x))})"
````

# Наследование

````python
from typing import Self

class Vector:
    def __init__(self, x: float, y: float) -> None:
        print('create Vector')
        self.x: float = x
        self.y: float = y

    def __add__(self, obj: Self) -> Self:
        # return Vector(self.x + obj.x, self.y + obj.y)
        # return type(self)(self.x + obj.x, self.y + obj.y)
        return self.__class__(self.x + obj.x, self.y + obj.y)

    def __repr__(self) -> str:
        return f"({self.x}, {self.y})"

class Circle(Vector):
    def __init__(self, x: float, y: float, r: float) -> None:
        print("create Circle")
        super().__init__(x, y)
        self.r: float  = r

    def get_center(self: Self) -> Vector:
        return Vector(self.x, self.y)

    def __add__(self: Self, obj: Self) -> Self:
        v: Vector = self.get_center() + obj.get_center()
        r:float = self.r + obj.r
        return type(self)(v.x, v.y, r)

    def __repr__(self) -> str:
        return f"Circle {self.r} {super().__repr__()}"
````

````python
c1 = Circle(10, 20, 5)
c2 = Circle(50, 40, 7)
c1
````

````python
c1 + c2
````

#### Множественное наследование как реализация интерфейсов

````python
class Golos:
    def scream(self):
        raise NotImplementedError()


class Eat:
    def eat(self):
        raise NotImplementedError()


class Go:
    def go(self):
        raise NotImplementedError()


class Cat(Golos, Eat, Go):
    def scream(self):
        return "Мяу"

    def eat(self):
        return "Кушаю"

    def go(self):
        return "Прыгаю"


class Dog(Golos, Eat, Go):
    def scream(self):
        return "Загрызу тебя!"

    def eat(self):
        return "Разгрызаю кости врагов!"

    def go(self):
        return "Несусть сшибая все на своем пути!"


class Turtle(Eat, Go):
    def eat(self):
        return "Загрызу тебя!"

    def go(self):
        return "Подниму левую ногу..."
````

````python
c = [Cat(), Dog(), Turtle()]
for el in c:
    print(el.__class__)
    if isinstance(el, Golos):
        print(el.scream())
    if isinstance(el, Eat):
        print(el.eat())
    if isinstance(el, Go):
        print(el.go())
````

#### Интерфейс Iterator

````python
class VectorDimensionsDoNotMatch(Exception):
    """
    Размерности векторов не совпадают
    """
    def __init__(self, l1, l2):
        message = f"Размерности векторов не одинаковы {l1} != {l2}"
        super().__init__(message)
        

class Vector:
    def __init__(self, *args):
        """
        x - координаты вектора
        """
        self.x = args
        self._index = 0

    def __add__(self: Self, obj: Self) -> Self:
        if len(self.x) != len(obj.x):
            raise VectorDimensionsDoNotMatch(len(self.x), len(obj.x))
        else:
            a = []
            for i in range(len(self.x)):
                a.append(self.x[i] + obj.x[i])
            return type(self)(*a)
            
    def __repr__(self: Self) -> str:
        return f"({", ".join(map(str, self.x))})"

    def __iter__(self):
        return self

    def __next__(self):
        if self._index < len(self.x):
            m = self.x[self._index]
            self._index += 1
            return m
        else:
            raise StopIteration()
````

````python
v1 = Vector(*[10, 20, 30, 40, 50, 60])
````

````python
for el in v1:
    print(el)
````

````python
class FibonacciIterator:
    def __init__(self, stop=10):
        self._stop = stop
        self._index = 0
        
        self._current = 0
        self._next = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self._index < self._stop:
            self._index += 1
            fib_number = self._current
            self._current, self._next = (
                self._next,
                self._current + self._next,
            )
            return fib_number
        else:
            raise StopIteration
````

````python
for el in FibonacciIterator(stop=20):
    print(el)
````

class Student:
    def __init__(self, name):

            return 0

        return sum(self.grades) / len(self.grades)
