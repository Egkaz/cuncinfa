```python
class Counter:
    def __init__(self, limit):
        self.limit = limit
        self._current = 0

    def __iter__(self):
        return self  # Итератор возвращает сам себя

    def __next__(self):
        if self._current >= self.limit:
            raise StopIteration
        result = self._current
        self._current += 1
        return result

    def reset(self):
        self._current = 0

# Использование
for number in Counter(3):
    print(number)  # Выведет: 0 1 2
```

```python
from typing import TypeVar, Iterator, Iterable

T = TypeVar('T')

class EmptyArray(Exception):
    ...

def search_min(a: Iterable[T]) -> T: 
    b: Iterator = iter(a)
    try:
        min_el: T = next(b)
    except StopIteration:
        raise EmptyArray()
        
    for el in b:
        if el < min_el:
            min_el = el
    return min_el
```

```python
a = [1,-1, 10, -20, 30]
search_min(a)
```

```python
class Parabola:
    def __init__(self, a: float, b: float, c: float):
        self.a: float = a
        self.b: float = b
        self.c: float = c

    def __call__(self, x: float) -> float:
        return self.a * x**2 + self.b * x + self.c

    def _D(self):
        return (self.b**2 - 4 * self.a * self.c)**(1/2)

    def get_sqrt(self):
        return (-self.b + self._D())/(2*self.a), \
                    (-self.b - self._D())/(2*self.a)

p1 = Parabola(-4, 3, 2)
p2 = Parabola(2, -4, 1)
p1(3), p2(5), p1.get_sqrt(), p2.get_sqrt()
```

```python
def search_extremum(a: Iterable[T], is_min=True) -> T: 
    def cmp1(el1: T, el2: T) -> bool:
        return el1 < el2

    def cmp2(el1: T, el2: T) -> bool:
        return el1 > el2

    def get_cmp(is_min: bool) -> Callable:
        if is_min:
            return cmp1
        else:
            return cmp2 
    a: Iterator = iter(a)
    try:
        min_el: T = next(a)
    except StopIteration:
        raise EmptyArray()
        
    for el in a:
        if get_cmp(is_min)(el, min_el):
            min_el = el
    return min_el
```

```python
class SearchExtremum:
    def __init__(self, is_min=True):
        # self.d = {
        #     True: self.cmp1,
        #     False: self.cmp2,
        # }
        # self.cmp = self.d[is_min]
        self.cmp = self.cmp1 if is_min else self.cmp2
        
        
    def cmp1(self, el1: T, el2: T) -> bool:
        return el1 < el2

    def cmp2(self, el1: T, el2: T) -> bool:
        return el1 > el2

    def __call__(self, a: Iterable[T])
        a: Iterator = iter(a)
        try:
            min_el: T = next(a)
        except StopIteration:
            raise EmptyArray()
            
        for el in a:
            if self.cmp(el, min_el):
                min_el = el
        return min_el
```

```python
class SearchUniversal:
    def __init__(self, f: Callable=min):
        self.cmp = f

    def __call__(self, a: Iterable[T]) -> T:
        a: Iterator = iter(a)
        try:
            min_el: T = next(a)
        except StopIteration:
            raise EmptyArray()
            
        for el in a:
            # if self.cmp(el, min_el):
            #    min_el = el
            min_el = self.cmp(el, min_el)
        return min_el
```

## Сортировка:
### Сортировка выбором (Selection Sort).

```python
def selection_sort(a: list[float]) -> None:
    class EmptyArray(Exception):
        ...
        
    def search_index_min(array: list[float], start=0) -> int:
        if len(array):
            min_el = array[start]
            min_i = start
            for i in range(start+1, len(array)):
                if array[i] < min_el:
                    min_el = array[i]
                    min_i = i
            return min_i
        else:
            raise EmptyArray()

    j = 0
    while j < len(a):
        min_i = search_index_min(a, start=j)
        a[j], a[min_i] = a[min_i], a[j]
        j += 1
    
```

```python
a = [9,81,1,2,-2,0]
selection_sort(a)
a
```

## Сортировка пузырьком (Bubble Sort) — классика для объяснения принципа обмена.

```python
def bubble_sort(a: list[float]) -> None:
    for i in range(len(a)-1):
        for j in range(len(a)-i-1):
            if a[j] > a[j+1]:
                a[j], a[j+1] = a[j+1], a[j]
```

```python
a = [9,81,1,2,-2,0]
bubble_sort(a)
a
```

## сортировка слиянием (MergeSort)

```python
def merge(a1: list[int], a2: list[int]) -> list[int]:
    b: list[int] = []
    i1: int = 0
    i2: int = 0
    while i1 < len(a1) and i2 < len(a2):
        if a1[i1] <= a2[i2]:
            b.append(a1[i1])
            i1 += 1
        else:
            b.append(a2[i2])
            i2 += 1

    while (i1 < len(a1)):
        b.append(a1[i1])
        i1 += 1
    while (i2 < len(a2)):
        b.append(a2[i2])
        i2 += 1
    return b
```

```python
merge([2,4,50,60], [1,2,3,4,5])
```

## Сортировка грузовых космических кораблей
Грузовой отдел стандартного космического корабля задается параметрами ширины, высоты и глубины. Так же каждый космических корабль по правилам империи должен возить с собой современную продвинутую аптечку (биокапсулу) на все случаи жизни, которая даже умеет из одной клетки восстанавливать целого человека. Размер такой аптечки зависит от фирмы и категории, но по сути тоже задается шириной, высотой и глубиной. 
Создайте классы 
- Класс Aptechka, который хранит ширину, высоту и глубину аптечки и умеет вычислять её объём.
- Класс Starship, который хранит размеры грузового отдела и аптечку, а также умеет вычислять оставшееся свободное пространство.
- Класс SortedDispetcher, который сортирует корабли по свободному пространству.
- Класс MergeDispetcher, который сливает несколько уже отсортированных потоков кораблей в один общий отсортированный поток.

Свободное пространство в грузовом отделе это пространство после вычета от туда размера аптечки.


Выполните следующий код для проверки задания:

```python
if __name__ == "__main__":
    a1 = Aptechka(20, 30, 10)
    a2 = Aptechka(10, 10, 10)
    a3 = Aptechka(10, 20, 10)

    s1 = Starship(100, 200, 300, a1)
    s2 = Starship(1000, 200, 300, a2)
    s3 = Starship(50, 50, 100, a3)

    ships = [s1, s2, s3]

    sd = SortedDispetcher()
    res = sd.sorted(ships)

    expected = [248000, 5994000, 59999000]

    assert [ship.free_space for ship in res] == expected

    md = MergeDispetcher()

    # Уже отсортированные потоки
    stream1 = [s3, s1]
    stream2 = [s2]

    res_merge = md.merge(stream1, stream2)

    assert [ship.free_space for ship in res_merge] == expected

    # Несколько потоков, включая пустой
    res_merge_many = md.merge([s3], [s1], [s2], [])

    assert [ship.free_space for ship in res_merge_many] == expected

    # Проверка слияния через вызов объекта
    res_call = md([s3], [s1], [s2])

    assert [ship.free_space for ship in res_call] == expected

    # Дополнительный пример, где потоки реально чередуются
    a4 = Aptechka(5, 5, 5)
    s4 = Starship(100, 100, 100, a4)

    # s4.free_space = 100 * 100 * 100 - 5 * 5 * 5 = 999875

    res_interleave = md.merge([s3, s1], [s4, s2])

    assert [ship.free_space for ship in res_interleave] == [
        248000,
        999875,
        5994000,
        59999000
    ]

    print("Все проверки выполнены")
```
