Эту ячейку студент запускает без изменений, кроме номера варианта.

```python
VARIANT = 1  # от 1 до 15


def make_contacts(v):
    names = [
        "Анна", "Борис", "Вера", "Глеб", "Дарья",
        "Егор", "Зоя", "Иван", "Кира", "Лев"
    ]

    def pretty(digits):
        return f"+{digits[0]} {digits[1:4]} {digits[4:7]}-{digits[7:9]}-{digits[9:11]}"

    def name(shift):
        return names[(v + shift) % len(names)]

    base_digits = f"7900111{20 + v:02d}{30 + v:02d}"
    base8 = "8" + base_digits[1:]
    base10 = base_digits[1:]

    tail_a = "79000009876"
    tail_b = "79000006798"

    contacts = [
        f"{name(0)}; {pretty(base_digits)}",
        f"{name(1)}; {pretty(base8)}",
        f"{name(2)}; {base10[:3]} {base10[3:6]}-{base10[6:8]}-{base10[8:10]}",
        f"{name(0)}; {pretty(base_digits)}",
        f"{name(3)}; {pretty(tail_a)}",
        f"{name(4)}; {pretty(tail_b)}",
        f"{name(5)}; +7 900 555-{v:02d}-{(v + 3) % 100:02d}",
        f"{name(6)}; +7 900 555-{v:02d}-{(v + 3) % 100:02d}",
        f"{name(7).upper()}; +7 900 777-{v:02d}-{(v + 1) % 100:02d}",
        "Строка без точки с запятой",
        "Ещё одна битая строка",
        f"{name(5)}; 12345",
        f"; +7 900 111-11-11",
        f"{name(6)}; 8-999-000-11-22",
    ]

    if v % 2 == 0:
        contacts.append("Дополнительная битая строка для чётного варианта")

    if v % 3 == 0:
        contacts.append(f"{name(8)}; 12-34")

    if v % 5 == 0:
        contacts.append(f"{name(9)}; 8-999-000-11-22")

    return contacts


def make_operations(v):
    payload = {
        'a': v,
        'b': v * 3,
        's': f'var{v}',
        v: 'numeric-key'
    }

    operations = [
        ('add', (v, 5), {}),
        ('add', (v,), {}),
        ('mul', (2, v), {}),
        ('div', (10, 0), {}),
        ('div', (v + 10, 2), {}),
        ('mod', (v, 0), {}),
        ('mod', (v + 7, 4), {}),
        ('int', (str(v + 1),), {}),
        ('int', ('bad',), {}),
        ('get', (payload, 'a'), {}),
        ('get', (payload, 'missing'), {}),
        ('get', (payload, v), {}),
        ('get', (payload, 's', 'default'), {}),
        ('concat', ('x', v, 'y'), {'sep': '-'}),
        ('concat', ('x', v), {'sep': v}),
        ('power', (2, 3), {}),
    ]

    if v % 2 == 0:
        operations.append(('mod', (1, 0), {}))

    if v % 3 == 0:
        operations.append(('unknown_extra', (), {}))

    if v % 5 == 0:
        operations.append(('concat', ('v', v), {'sep': '='}))

    return operations, payload


CONTACTS = make_contacts(VARIANT)
OPERATIONS, PAYLOAD = make_operations(VARIANT)
```

После выполнения этой ячейки доступны:

## Какие функции нужно реализовать

```python
def op_add(a, b):
    return a + b


def op_mul(a, b):
    return a * b


def op_div(a, b):
    return a // b


def op_mod(a, b):
    return a % b


def op_int(value):
    return int(value)


def op_concat(*parts, sep=''):
    if not isinstance(sep, str):
        raise TypeError('sep must be str')
    return sep.join(str(part) for part in parts)


def op_get(data, key):
    return data[key]
```

# 8. Эталонное решение для преподавателя

Ниже — код, который можно использовать для проверки. Студентам его показывать не нужно.

Этот код предполагает, что функции `make_contacts` и `make_operations` уже определены из ячейки с данными.

```python
import pickle


def normalize_phone(raw):
    digits = ''.join(ch for ch in raw if ch.isdigit())

    if len(digits) == 10 and digits.startswith('9'):
        digits = '7' + digits
    elif len(digits) == 11 and digits.startswith('8'):
        digits = '7' + digits[1:]

    if len(digits) == 11 and digits.startswith('7'):
        return digits

    return None


def solve_phonebook(contacts):
    broken = 0
    invalid = 0
    raw_pairs = []

    for line in contacts:
        if ';' not in line:
            broken += 1
            continue

        name_raw, phone_raw = line.split(';', 1)

        name = name_raw.strip().lower()
        phone = normalize_phone(phone_raw)

        if not name or phone is None:
            invalid += 1
            continue

        raw_pairs.append((name, phone))

    unique_pairs = list(set(raw_pairs))
    unique_pairs.sort()

    by_phone = {}
    for name, phone in unique_pairs:
        by_phone.setdefault(phone, set()).add(name)

    tail_groups = {}
    for phone in by_phone:
        tail_groups.setdefault(frozenset(phone[-4:]), set()).add(phone)

    shared_phones = sum(1 for names in by_phone.values() if len(names) >= 2)
    tail_collisions = sum(1 for phones in tail_groups.values() if len(phones) >= 2)
    unique_names = len({name for name, phone in unique_pairs})
    bad = broken + invalid

    phone_checksum = sum(int(ch) for phone in by_phone for ch in phone) % 1000

    return (
        shared_phones * 1_000_000
        + tail_collisions * 100_000
        + unique_names * 10_000
        + bad * 1_000
        + phone_checksum
    )


def op_add(a, b):
    return a + b


def op_mul(a, b):
    return a * b


def op_div(a, b):
    return a // b


def op_mod(a, b):
    return a % b


def op_int(value):
    return int(value)


def op_concat(*parts, sep=''):
    if not isinstance(sep, str):
        raise TypeError('sep must be str')
    return sep.join(str(part) for part in parts)


def op_get(data, key):
    return data[key]


def solve_dispatcher(operations, variant):
    dispatch = {
        'add': op_add,
        'mul': op_mul,
        'div': op_div,
        'mod': op_mod,
        'int': op_int,
        'concat': op_concat,
        'get': op_get,
    }

    counters = {
        'ok': 0,
        'unknown': 0,
        'zero': 0,
        'key': 0,
        'type': 0,
        'error': 0,
    }

    results = []

    for op_name, args, kwargs in operations:
        func = dispatch.get(op_name)

        if func is None:
            counters['unknown'] += 1
            continue

        try:
            result = func(*args, **kwargs)
        except ZeroDivisionError:
            counters['zero'] += 1
        except KeyError:
            counters['key'] += 1
        except (TypeError, ValueError):
            counters['type'] += 1
        except Exception:
            counters['error'] += 1
        else:
            counters['ok'] += 1
            results.append(result)

    numeric_sum = sum(
        x for x in results
        if isinstance(x, int) and not isinstance(x, bool)
    )

    text_len = sum(
        len(x) for x in results
        if isinstance(x, str)
    )

    data = counters.copy()
    data['numeric_sum'] = numeric_sum
    data['text_len'] = text_len

    filename = f"variant_{variant}.pickle"

    with open(filename, 'wb') as f:
        pickle.dump(data, f)

    with open(filename, 'rb') as f:
        loaded = pickle.load(f)

    errors_total = (
        loaded['unknown']
        + loaded['zero']
        + loaded['key']
        + loaded['type']
        + loaded['error']
    )

    return loaded['numeric_sum'] * 1000 + loaded['text_len'] * 10 + errors_total
```
