# Генераторы

```python
a: list[int] = [1,2,3,4,5]
b = []
for el in a:
    b.append(el**2)
b
```

```python
b = [el**2 for el in a]
b
```

```python
def d(el):
    return el**2

b = [d(el) for el in a]
b
```

```python
type(b)
```

```python
b = (el**2 for el in a)
b
```

```python
type(b)
```

```python
for el in b:
    print(el)
```
