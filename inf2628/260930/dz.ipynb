## Часть 1. Поиск второго минимума

На уроке 260928 вы искали второй минимум. Обратите внимание на нюансы:

- **Два разных определения.** «Второй минимум с учётом повторов»: в `[1, 1, 3]` ответ `1` (второй элемент отсортированного массива). «Второй минимум среди различных значений»: в `[1, 1, 3]` ответ `3`. Всегда читайте условие!

- **Граничные случаи.** Пустой массив, один элемент, все элементы равны — решите, что возвращать (в наших задачах — `None` либо исключение).

````python
# === Задача 1.1 (1 балл). Второй минимум за два прохода ===
#
# Реализуйте функцию двумя проходами по массиву, БЕЗ сортировки и встроенных
# min/sorted. В этой задаче гарантируется, что все элементы массива различны.
# Если в массиве меньше двух элементов, верните None.
#
# Идея: первым проходом найдите минимум и запомните его ИНДЕКС,
# вторым проходом ищите минимум среди всех элементов, кроме найденного.

def second_min_two_pass(arr):
    '''Возвращает второй минимум массива с различными элементами за два прохода.
    Если элементов меньше двух, возвращает None.'''
    if len(arr) < 2:
        return None
    m1 = None
    m2 = None
    for i in arr:
        if m1 is None or i < m1:
            m1, m2 = i, m1
        elif m2 is None or (i < m2 and i != m1):
            m2 = i
    return m2
    


# --- Тесты (не менять!) ---
assert second_min_two_pass([3, 1, 2]) == 2
assert second_min_two_pass([5, 4, 3, 2, 1]) == 2
assert second_min_two_pass([2, 1]) == 2
assert second_min_two_pass([10, 20, 30]) == 20
assert second_min_two_pass([-10, -5, 0]) == -5
assert second_min_two_pass([7]) is None
assert second_min_two_pass([]) is None
assert second_min_two_pass([1, 1, 3]) == 3
print('Задача 1.1: тесты пройдены')
````

```python
class NthMinFinder:
    def __init__(self, n):
        self.n = n
        self.mins = []          # отсортированный список из <= n наименьших элементов

    def _insert_sorted(self, x, keep_size=True):
        '''Вставляет x в self.mins с сохранением порядка;
        при включённом ограничении размера отбрасывает последний элемент.'''
        i = 0
        while i < len(self.mins) and self.mins[i] <= x:
            i += 1
        self.mins.insert(i, x)
        if keep_size and len(self.mins) > self.n:
            self.mins.pop()

    def add(self, x):
        if len(self.mins) < self.n:
            self._insert_sorted(x, keep_size=False)
        elif x < self.mins[-1]:        # x меньше наибольшего из хранимых — есть смысл вставлять
            self._insert_sorted(x)

    def result(self):
        if len(self.mins) < self.n:
            return None
        return self.mins[-1]           # наибольший из топ-наименьших и есть n-й минимум

    def __call__(self, arr):
        self.mins = []                 # сбрасываем состояние перед новым прогоном
        for x in arr:
            self.add(x)
        return self.result()
```

**Разбор.**

- Ключевой инвариант: `self.mins` всегда отсортирован и содержит не более `n` элементов. Тогда наибольший из них — в точности n-й минимум среди просмотренных.
- Если список ещё не заполнен — просто вставляем. Если заполнен — новый элемент имеет смысл вставлять, только когда он строго меньше наибольшего хранимого.
- `__call__` — магический метод, делающий объект вызываемым (проходили на уроке 260923).
- Сложность: `add` — `O(n)`, на потоке из `k` элементов — `O(k·n)`; память — `O(n)`. Для небольших фиксированных `n` это быстрее полной сортировки.

```python
def nth_min_counting(arr, n):
    if n < 1 or n > len(arr):
        return None
    counts = {}                          # значение -> сколько раз встретилось
    for x in arr:
        counts[x] = counts.get(x, 0) + 1
    seen = 0                             # сколько позиций уже «занято»
    for key in sorted(counts):           # идём по значениям по возрастанию
        seen += counts[key]
        if seen >= n:                    # n-я позиция попала в блок этого значения
            return key
    return None                          # недостижимо при корректном n
```

```python
def binary_search(arr, target):
    left, right = 0, len(arr) - 1     # границы интервала поиска
    while left <= right:              # пока интервал непустой
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:       # искомый элемент правее середины
            left = mid + 1
        else:                         # искомый элемент левее середины
            right = mid - 1
    return -1                         # интервал пуст — элемента нет в массиве
```

**Разбор.**

- Инвариант: если элемент есть в массиве, он всегда находится в `[left, right]`. Обновления границ через `mid ± 1` инвариант не нарушают, потому что сам `mid` уже проверен.

```python
    left, right = 0, len(arr)         # инвариант: ответ лежит в [left, right]
```

```python
def sqrt_bisect(a, eps=1e-10):
    if a < 0:
        raise ValueError('квадратный корень из отрицательного числа не определён')
    lo, hi = 0.0, max(1.0, float(a))   # корень всегда лежит на [0, max(1, a)]
    while hi - lo > 2 * eps:
        mid = (lo + hi) / 2
        if mid * mid <= a:
            lo = mid                   # mid^2 <= a: корень не левее середины
        else:
            hi = mid                   # mid^2 > a: корень левее середины
    return (lo + hi) / 2
```

**Разбор.**

```python
def bisect_root(f, lo, hi, eps=1e-10, max_iter=1000):
    f_lo = f(lo)
    if f_lo * f(hi) > 0:
        raise ValueError('на концах отрезка функция должна менять знак')
    for _ in range(max_iter):
        if hi - lo <= 2 * eps:          # отрезок стал короче требуемой точности — стоп
            break
        mid = (lo + hi) / 2
        f_mid = f(mid)
        if f_lo * f_mid <= 0:           # смена знака на левой половине — корень там
            hi = mid
        else:                           # иначе корень на правой половине
            lo = mid
            f_lo = f_mid                # запоминаем значение на новом левом конце (экономим вызовы f)
    return (lo + hi) / 2
```

**Разбор.**

- Инвариант: на концах текущего отрезка функция имеет разные знаки, поэтому корень всегда внутри.
- За итерацию длина отрезка уменьшается вдвое; середина финального отрезка отличается от корня не больше чем на `eps`.
- `max_iter` — защита от зависания при слишком маленьком `eps`.
- Исключение при отсутствии смены знака делает функцию безопасной: молча вернуть «корень», которого нет, — худший сценарий.

```python
def stable_counting_sort(pairs, k):
    counts = [0] * k
    for key, _ in pairs:
        counts[key] += 1

    # стартовые позиции в результате для каждого ключа (префиксные суммы счётчиков)
    starts = [0] * k
    total = 0
    for key in range(k):
        starts[key] = total
        total += counts[key]

    result = [None] * len(pairs)
    pos = starts[:]                    # текущая свободная позиция для каждого ключа
    for key, value in pairs:           # обход слева направо = устойчивость!
        result[pos[key]] = (key, value)
        pos[key] += 1
    return result
```

```python
def merge_two(a, b):
    result = []
    i = j = 0
    while i < len(a) and j < len(b):
        if a[i] <= b[j]:        # <= сохраняет устойчивость: равные из a идут раньше
            result.append(a[i])
            i += 1
        else:
            result.append(b[j])
            j += 1
    result.extend(a[i:])        # хвост того списка, который не закончился (второй уже пуст)
    result.extend(b[j:])
    return result
```

```python
def count_inversions(arr):
    def sort_count(a):
        '''Возвращает пару: (отсортированная копия a, число инверсий в a).'''
        n = len(a)
        if n <= 1:
            return a[:], 0
        mid = n // 2
        left, inv_left = sort_count(a[:mid])
        right, inv_right = sort_count(a[mid:])
        merged = []
        inv = inv_left + inv_right       # инверсии внутри половин уже посчитаны рекурсией
        i = j = 0
        while i < len(left) and j < len(right):
            if left[i] <= right[j]:
                merged.append(left[i])
                i += 1
            else:
                merged.append(right[j])
                j += 1
                inv += len(left) - i     # right[j] меньше ВСЕХ оставшихся в левой половине!
        merged.extend(left[i:])
        merged.extend(right[j:])
        return merged, inv

    return sort_count(arr)[1]
```

**Разбор.**

- Любая инверсия либо целиком внутри левой половины, либо целиком внутри правой (считает рекурсия), либо «пересекает» разрез (считаем при слиянии). Три случая не пересекаются — значит, ничего не посчитано дважды и ничего не пропущено.
- Ключевой момент: когда в слиянии берём элемент из правой половины, он меньше всех ещё не взятых элементов левой половины (обе половины отсортированы), поэтому прибавляем `len(left) − i`.
- Сложность `O(n log n)` вместо `O(n²)` у наивного подсчёта.

## Часть 7. Теория вероятности: комбинаторика и классическое определение вероятности

Напоминание определений (по Гнеденко):

- **Элементарное событие** (исход) ω — неделимый результат случайного эксперимента.
- **Пространство элементарных событий** Ω — множество всех возможных исходов.
- **Событие** — любое подмножество A ⊆ Ω.
- Если все исходы равновозможны, **классическое определение вероятности**: `P(A) = |A| / |Ω|` — число исходов, входящих в событие, делённое на число всех исходов.
- События **несовместны**, если у них нет общих элементарных исходов (`A ∩ B = ∅`).
- Группа событий образует **полную группу**, если события попарно несовместны и их объединение равно Ω: в каждом эксперименте происходит ровно одно из них.
- **Невозможное событие** ∅ имеет вероятность 0, **достоверное событие** Ω — вероятность 1.
- Комбинаторика: размещения `A(n, k) = n! / (n − k)!` (порядок важен), сочетания `C(n, k) = n! / (k! (n − k)!)` (порядок не важен).

```python
def factorial(n):
    if n < 0:
        raise ValueError('факториал определён только для неотрицательных чисел')
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result                        # при n = 0 цикл не выполняется: 0! = 1 автоматически

def permutations(n, k):
    # A(n, k) = n! / (n - k)!: выбираем и УПОРЯДОЧИВАЕМ k элементов из n
    return factorial(n) // factorial(n - k)

def combinations(n, k):
    # C(n, k) = n! / (k! (n - k)!): выбираем БЕЗ учёта порядка
    return factorial(n) // (factorial(k) * factorial(n - k))
```

```python
        return self.sorted_prices()[n - 1]      # n-й минимум в отсортированном списке — просто индекс!

        s = self.sorted_prices()
        # число элементов в [lo, hi] = (число элементов <= hi) − (число элементов < lo)

        for x in self.prices:
            counts[x] = counts.get(x, 0) + 1
        best, best_cnt = None, -1       # идём по ключам ПО ВОЗРАСТАНИЮ:
        for key in sorted(counts):      # при равенстве частот строгое > не перезапишет меньшую цену!
            if counts[key] > best_cnt:
                best, best_cnt = key, counts[key]
        return best

    def optimal_price(self, a, b, cost):
        # Прибыль π(p) = (p − cost)(a − b·p). Раскрываем скобки:
        # π(p) = −b·p² + (a + b·cost)·p − a·cost — парабола ветвями вниз.
        # Производная π'(p) = a + b·cost − 2·b·p — линейная УБЫВАЮЩАЯ функция.

        # На концах [cost, a/b]: π'(cost) = a − b·cost > 0, π'(a/b) = b·cost − a < 0 → знак меняется.
        def dp(p):
            return a + b * cost - 2 * b * p
        return bisect_root(dp, float(cost), a / b, eps=1e-9)   # задача 4.2!
```

```python
import random


class DiceExperiment:
    def __init__(self, n_dice=2, faces=6, seed=None):
        self.n_dice = n_dice
        self.faces = faces
        self.rng = random.Random(seed)   # свой генератор: воспроизводимость, глобальный random не трогаем!
        self._omega = None               # кэш пространства элементарных событий (ленивая инициализация)

    def sample_space(self):
        if self._omega is None:
            outcomes = [()]              # начинаем с одного ПУСТОГО исхода: к нему будем дописывать кубики!
            for _ in range(self.n_dice):
                # декартово произведение: к каждому текущему исходу приписываем все значения новой грани кубика!
                outcomes = [o + (face,) for o in outcomes for face in range(1, self.faces + 1)]
            self._omega = set(outcomes)  # Ω — множество кортежей, как в теории вероятностей!
        return self._omega               # кэш: повторные вызовы не пересчитывают!

    def theoretical_probability(self, predicate):
        omega = self.sample_space()
        event = {e for e in omega if predicate(e)}    # событие — ПОДМНОЖЕСТВО Ω: генератор множества по признаку!
        return len(event) / len(omega)                # классическое определение Гнеденко: |событие| / |Ω|!

    def simulate(self, n_trials, predicate):
        hits = 0                          # счётчик благоприятных исходов моделирования!
        for _ in range(n_trials):
            outcome = tuple(self.rng.randint(1, self.faces) for _ in range(self.n_dice))  # бросок: кортеж как в Ω!
            if predicate(outcome):        # тот же предикат, что и в теории — единый интерфейс!
                hits += 1
        return hits / n_trials            # эмпирическая вероятность = доля благоприятных исходов!

    def sum_distribution(self):
        omega = self.sample_space()
        counts = {}                       # сумма → частота: подсчёт как в сортировке подсчётом через словарь!
        for e in omega:
            s = sum(e)
            counts[s] = counts.get(s, 0) + 1
        return {s: counts[s] / len(omega) for s in sorted(counts)}   # ключи ПО ВОЗРАСТАНИЮ — как требует тест!

    def median_sum(self):
        sums = [sum(e) for e in self.sample_space()]   # все суммы исходов Ω С ПОВТОРАМИ!
        k = (len(sums) + 1) // 2          # номер «нижней» медианы при 1-индексации!
        return NthMinFinder(k)(sums)      # переиспользуем класс из задачи 2.2 — поиск k-го минимума!

    def check_sum_group(self):
        omega = self.sample_space()
        possible = sorted({sum(e) for e in omega})    # все возможные значения суммы (для двух кубиков: 2..12)!
        events = [{e for e in omega if sum(e) == s} for s in possible]   # событие «сумма равна s» для каждого s!
        return is_complete_group(events, omega)       # задача 7.3: попарная несовместность + покрытие Ω!

    def __repr__(self):
        return f'DiceExperiment(n_dice={self.n_dice}, faces={self.faces})'
```

**Разбор.**

- `sample_space` строит декартово произведение граней: на каждом шаге к каждому уже построенному исходу приписываются все значения нового кубика. Для двух кубиков получается 6, затем 36 исходов — это и есть пространство элементарных событий, о котором пишет Гнеденко. Результат кэшируется.
- Теоретическая вероятность считается буквально по определению: событие — подмножество Ω, вероятность — отношение мощностей. Именно поэтому удобно хранить Ω как множество кортежей.
- `simulate` использует собственный генератор `random.Random(seed)`: эксперимент воспроизводим (тест с `seed=42` всегда проходит), а глобальный `random` не «портится».
- `median_sum` сводится к поиску k-го минимума среди всех сумм исходов (с повторами) — переиспользуем класс из задачи 2.2. Для двух кубиков: всего 36 сумм, «нижняя» медиана — 18-й минимум, он равен 7 (проверьте: 15 сумм ≤ 6 и 21 сумма ≤ 7).
- `check_sum_group` проверяет теоретико-вероятностный факт: события «сумма равна s» попарно несовместны и покрывают Ω, поэтому их вероятности в сумме дают ровно 1 — это проверено и в тестах ячейки.
- Тест на три кубика: исходов `6³ = 216`; число способов получить сумму 8 тремя слагаемыми от 1 до 6 равно `C(7, 2) = 21` (метод «звёзды и полосы»: после замены переменной решаем в неотрицательных целых).
