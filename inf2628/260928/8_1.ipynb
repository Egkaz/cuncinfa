```python
DEBUG = False

def dprint(*args, **kwargs):
    if DEBUG:
        print(*args, **kwargs)


class SearchNMin:
    def __init__(self, n: int) -> None:
        self.n: int = n
        self.m_min: list[float|None] = [None] * self.n

    def _shift(self, j: int) -> None:
        for i in range(len(self.m_min)-1, j, -1):
            self.m_min[i] = self.m_min[i-1]

    def __call__(self, a: list[float]) -> float:
        dprint(a)
        self.m_min:list[float] = sorted(a[:self.n])
        dprint(self.m_min)
        for j in range(self.n, len(a)):
            dprint(a[j])
            for i, m in enumerate(self.m_min):
                dprint(i, m)
                if a[j] < m:
                    self._shift(i)
                    self.m_min[i] = a[j]
                    dprint(self.m_min)
                    break
                    
        return self.m_min[-1]


s1 = SearchNMin(1)
s2 = SearchNMin(2)
s5 = SearchNMin(5)
b = [1]
c = [1,1,1,1,1]
d = [1,2,3,4,4]
# print(s1(a))
assert s1(a) == 0
assert s2(a) == 1
assert s5(a) == 4
assert s1(b) == 1
assert s5(c) == 1
assert s5(d) == 4
```

Реализовать сортировку InsertionSort

```python
def merge(a: list[int], i11:int, i12:int, i21:int, i22:int) -> None:
    """
    i11 - индекс первого элемента в первом массиве
    i21 - индекс первого элемента во втором массиве
    """
    # i11, i12, i21, i22 = min(i11, i21), max(i11, i21), min(i21, i22), max(i21, i22)
    i11, i12, i21, i22 = sorted([i11, i12, i21, i22])
    i1 = i11
    i2 = i21
    b = []
    while i1 <= i12 and i2 <= i22:
        if a[i1] <= a[i2]:
            b.append(a[i1])
            i1 += 1
        else:
            b.append(a[i2])
            i2 += 1

    if i11 < i21: # первый массив лежит левее второго
        if i1 > i12: #  закончился левый массив
            a[i11:i1+i2] = b
        else: # i2 > i22  # закончился правый массив
            b.extends(a[i1:i12+1])
            # b += a[i1:i12+1]
            a[i11:i22+1] = b
```

```python
def merge(a1: list[int], a2: list[int]) -> list[int]:
    b: list[int] = []
    i1: int = 0
    i2: int = 0
    while i1 < len(a1) and i2 < len(a2):
        if a1[i1] <= a2[i2]:
            b.append(a1[i1])
            i1 += 1
        else:
            b.append(a2[i2])
            i2 += 1

    b.extend(a1[i1:])
    b.extend(a2[i2:])

    return b
```
