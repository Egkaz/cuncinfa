### Теория: первый код, переменные, типы данных, ввод и вывод

> **Основы синтаксиса**
> - `print()` выводит данные на экран.
> - `#` начинает комментарий.
> - Переменная хранит значение: `name = "Аня"`.
> - Основные типы данных: `int`, `float`, `str`, `bool`.
> - `input()` всегда возвращает строку.
> - Чтобы получить число из ввода, используй `int(input(...))` или `float(input(...))`.

Арифметические операторы:

- `+` — сложение;
- `-` — вычитание;
- `*` — умножение;
- `/` — деление;
- `//` — целая часть от деления;
- `%` — остаток от деления;
- `**` — степень.

Строки:

- строки можно складывать: `"a" + "b"`;
- строки можно повторять: `"ha" * 3`;
- `f-строки` позволяют подставлять переменные: `f"Привет, {name}"`.

````python
# Пример 1. Первая программа
print("Привет, Python!")
print("Я изучаю основы синтаксиса")
````

````python
# Пример 2. Переменные
name: str = "Маша"
age: int = 14

print("Привет, меня зовут", name)
print(f"Мне {age} лет")
````

````python
# Пример 3. Типы данных
number: int = 10
price:float = 99.99
message: str = "Привет"
is_student: boolean = True

print(type(number))
print(type(price))
print(type(message))
print(type(is_student))
````

````python
# Пример 4. Ввод данных
name = input("Как тебя зовут? ")
age = int(input("Сколько тебе лет? "))

print("Привет,", name)
print("Через 10 лет тебе будет", age + 10)
````

````python
# Пример 5. Арифметика
a = 20
b = 6

print("a + b =", a + b)
print("a - b =", a - b)
print("a * b =", a * b)
print("a / b =", a / b)
print("a // b =", a // b)
print("a % b =", a % b)
print("a ** b =", a ** b)
````

````python
# Пример 6. Строки и f-строки
language: str = "Python"
lesson: str = "урок"

print(language + " " + lesson)
print(language * 3)

name: str = "Иван"
score: int = 95

print(f"Ученик {name} получил {score} баллов")
````

````python
# Пример 7. Мини-анкета
name: str = input("Как тебя зовут? ")
age: int = int(input("Сколько тебе лет? "))
city: str = input("Из какого ты города? ")

print("Анкета заполнена!")
print(f"Имя: {name}")
print(f"Возраст: {age}")
print(f"Город: {city}")
````
